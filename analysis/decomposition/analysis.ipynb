```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm

# 查找系统中的中文字体
font_path = fm.findfont(fm.FontProperties(family='SimSun'))
print(f"Using font: {font_path}")

# 设置Matplotlib的字体
plt.rcParams['font.sans-serif'] = ['SimSun'] # 使用文泉驿微米黑字体
plt.rcParams['axes.unicode_minus'] = False # 解决负号 '-' 显示为方块的问题

# 创建一些数据

# 读取CSV文件
df = pd.read_csv('series.csv')

# 检查列名是否符合预期
if 'original' not in df.columns or 'seasonal' not in df.columns or 'trend' not in df.columns:
    raise ValueError("CSV文件中缺少必要的列: 'original', 'seasonal', 或 'trend'")

# 设置绘图大小
plt.figure(figsize=(14, 8))

# 绘制原始序列
plt.plot(df['original'], label='原始值', color='red')

# 绘制季节性成分
plt.plot(df['seasonal'], label='波动项', color='green')

# 绘制趋势成分
plt.plot(df['trend'], label='长期趋势项', color='blue')

# 添加标题和标签
plt.title('风速序列趋势分解')
plt.xlabel('时间')
plt.ylabel('风速值')

# 添加图例
plt.legend()

# 显示图表
plt.tight_layout()
plt.show()
```
