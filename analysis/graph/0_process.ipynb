```python
import os
import pandas as pd
from tqdm import tqdm

def process_dated_csvs():
    # 获取当前目录下所有dated_*.csv文件
    csv_files = [f for f in os.listdir() if f.startswith('dated_') and f.endswith('.csv')]
    
    if not csv_files:
        print("没有找到dated_*.csv文件")
        return
    
    print(f"找到{len(csv_files)}个dated_*.csv文件，开始处理...")
    
    for file in tqdm(csv_files, desc="处理文件中"):
        try:
            # 读取CSV文件
            df = pd.read_csv(file)
            
            # 检查并删除Date列（不区分大小写）
            cols_to_drop = [col for col in df.columns if col == 'Date']
            
            if cols_to_drop:
                df.drop(columns=cols_to_drop, inplace=True)
                # 保存修改后的文件（覆盖原文件）
                df.to_csv(file, index=False)
                print(f"\n已处理: {file} - 删除了{len(cols_to_drop)}个Date列")
            else:
                print(f"\n跳过: {file} - 没有Date列")
                
        except Exception as e:
            print(f"\n处理文件{file}时出错: {str(e)}")
    
    print("\n所有文件处理完成！")

if __name__ == "__main__":
    process_dated_csvs()
```
